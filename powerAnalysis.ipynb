```python
import numpy as np
from statsmodels.stats.power import TTestPower

### this is going to be the power analysis for the 5k timings
print('\n5K Time Power Analysis\n\n')

############################# - difference between conditions
# this difference is based on the difference that was seen in the nike shoe nyt article (4%)
origdiffmean = (20 * 60) * 0.04  # in this case we might estimate the difference that we could see or detect...
############################# - SD of the differences
# this is going to be the standard deviation of the 5k timing differences that we see in the data
# I got this value in 5ktimings.xlsx, based on data of repeat 5k runs from parkrun data. 
origdiffstd = 15.65

############################# - Power computations
effect_size = origdiffmean / origdiffstd
alpha = 0.05/3 # Bonferroni correction for multiple comparisons
power = 0.9
alternative = 'two-sided' ## what does this mean???
# initialize the power analysis
analysis = TTestPower()
# calc sample size
sample_size = analysis.solve_power(effect_size=effect_size, alpha=alpha, power=power, alternative=alternative)
print('N min = %f' % sample_size)


############################# - sweeping method to see some contrasts
print('\n\n\nHere is another method for comparing the sample sizes for different SD, and different expected changes between conditions\n')
#### this is what chatgpt made for comparison
import numpy as np
import statsmodels.stats.power as smp

# Assumptions
baseline_time = 18 * 60  # 20-minute 5K in seconds  # 25-minute 5K in seconds
expected_percentage_change = np.array([0.03, 0.04, 0.05, 0.1])  # 3% to 5% change from previous device's CI
expected_reduction = baseline_time * expected_percentage_change  # Convert percentage change to absolute time reduction
std_dev_values = [15.65, 30, 60]  # Testing different standard deviations  # Assumed standard deviation from literature
alpha = 0.05  # Significance level

# Compute required sample size for 80% and 90% power
powers = [0.8, 0.9]
sample_sizes = {sd: {} for sd in std_dev_values}

for sd in std_dev_values:
    for power in powers:
        sample_sizes[sd][power] = []
        for diff in expected_reduction:
            effect_size = diff / sd  # Cohen's dz for paired t-test
            n = smp.TTestPower().solve_power(effect_size, alpha=alpha, power=power, alternative='two-sided')
            sample_sizes[sd][power].append(int(np.ceil(n)))

# Display results
for sd, power_dict in sample_sizes.items():
    print(f"\nStandard Deviation: {sd}s")
    for power, sizes in power_dict.items():
      print(f"Required sample size for {int(power * 100)}% power:")
      for diff, size in zip(expected_reduction, sizes):
        print(f"  Detecting {diff:.1f}s change ({expected_percentage_change[np.where(expected_reduction == diff)[0][0]] * 100:.0f}% at 20 min): {size} participants")

```
